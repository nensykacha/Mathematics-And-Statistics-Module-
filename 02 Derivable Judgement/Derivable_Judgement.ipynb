### Data 

```python
import pandas as pd
import numpy as np
import uuid
from datetime import datetime, timedelta

# Number of records
n = 100

# Random seed for reproducible results
np.random.seed(42)

# Categories
age_groups = ["18-25", "26-35", "36-45", "46-60", "60+"]
genders = ["Male", "Female", "Other"]
regions = ["North", "South", "East", "West"]
smoking_statuses = ["Smoker", "Non-Smoker", "Former Smoker"]
exercise_frequencies = ["Daily", "Weekly", "Rarely", "Never"]

# Generate Age
ages = np.random.randint(18, 81, n)

def get_age_group(age):
    if 18 <= age <= 25:
        return "18-25"
    elif 26 <= age <= 35:
        return "26-35"
    elif 36 <= age <= 45:
        return "36-45"
    elif 46 <= age <= 60:
        return "46-60"
    else:
        return "60+"

age_group = [get_age_group(age) for age in ages]

# Generate Weight
weights = np.random.randint(45, 101, n)


# Generate Gender
gender = np.random.choice(
    genders,
    n,
    p=[0.48, 0.48, 0.04]
)


# Generate Region
region = np.random.choice(
    regions,
    n
)


# Generate Smoking Status
smoking_status = np.random.choice(
    smoking_statuses,
    n,
    p=[0.20, 0.60, 0.20]
)


# Generate Exercise Frequency
exercise_frequency = np.random.choice(
    exercise_frequencies,
    n,
    p=[0.25, 0.40, 0.25, 0.10]
)


# Generate Height
# Height is generated internally only
# to calculate realistic BMI
heights = np.random.uniform(1.50, 1.90, n)

# BMI = weight / height^2
bmi = weights / (heights ** 2)

# Round BMI
bmi = np.round(bmi, 2)


# Blood Pressure
blood_pressure = np.random.normal(125, 18, n)
blood_pressure = np.clip(blood_pressure, 90, 190)
blood_pressure = np.round(blood_pressure, 1)


# Diabetes
diabetes = np.random.choice(
    [True, False],
    n,
    p=[0.15, 0.85]
)


# Hypertension
hypertension = np.random.choice(
    [True, False],
    n,
    p=[0.25, 0.75]
)

# Cholesterol Level
cholesterol_level = np.random.normal(195, 35, n)
cholesterol_level = np.clip(cholesterol_level, 120, 320)
cholesterol_level = np.round(cholesterol_level, 1)

# Glucose Level
glucose_level = np.random.normal(100, 20, n)
glucose_level = np.clip(glucose_level, 70, 220)
glucose_level = np.round(glucose_level, 1)


# Visit Date
start_date = datetime(2024, 1, 1)
end_date = datetime(2026, 9, 27)

date_range = (end_date - start_date).days

visit_dates = [
    start_date + timedelta(
        days=np.random.randint(0, date_range + 1)
    )
    for _ in range(n)
]

visit_dates = [
    date.strftime("%Y-%m-%d")
    for date in visit_dates
]

# Record ID
record_id = [str(uuid.uuid4()) for _ in range(n)]


# Create DataFrame
df = pd.DataFrame({
    "record_id": record_id,
    "age_group": age_group,
    "age": ages,
    "weight": weights,
    "gender": gender,
    "region": region,
    "smoking_status": smoking_status,
    "exercise_frequency": exercise_frequency,
    "bmi": bmi,
    "blood_pressure": blood_pressure,
    "diabetes": diabetes,
    "hypertension": hypertension,
    "cholesterol_level": cholesterol_level,
    "glucose_level": glucose_level,
    "visit_date": visit_dates
})

# Display Dataset
print("Health Dataset:")
print(df)

# Dataset Information
print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
print(df.info())

# Save CSV
df.to_csv("health_records.csv", index=False)

print("\nCSV file created successfully: health_records.csv")
```

### 2. Calculate Confidence Intervals for key numerical data like age, weight, etc

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("health_records.csv")

numerical_columns = [
    "age",
    "weight",
    "bmi",
    "blood_pressure",
    "cholesterol_level",
    "glucose_level"
]

# We are using a 95% confidence interval
confidence_level = 0.95

# Significance level
alpha = 1 - confidence_level

# CALCULATE CONFIDENCE INTERVAL
print("\n")
print("=" * 75)
print("95% CONFIDENCE INTERVALS")
print("=" * 75)

# Loop through every numerical column
for column in numerical_columns:

    # Remove missing values, if any
    data = df[column].dropna()

    # Calculate sample size
    n = len(data)

    # Calculate mean
    mean = np.mean(data)

    # Calculate standard deviation
    std = np.std(data, ddof=1)

    # Calculate Standard Error
    standard_error = std / np.sqrt(n)

    # Calculate t critical value
    # df = n - 1
    t_critical = stats.t.ppf(
        1 - alpha / 2,
        df=n - 1
    )

    # Calculate margin of error
    margin_of_error = t_critical * standard_error

    # Calculate lower confidence limit
    lower_limit = mean - margin_of_error

    # Calculate upper confidence limit
    upper_limit = mean + margin_of_error

    # Print results
    print("\nVariable:", column)
    print("Sample Size:", n)
    print("Mean:", round(mean, 2))
    print("Standard Deviation:", round(std, 2))
    print("Standard Error:", round(standard_error, 2))
    print("Critical t-value:", round(t_critical, 3))
    print("Margin of Error:", round(margin_of_error, 2))
    print("95% Confidence Interval:",
          "(",
          round(lower_limit, 2),
          ",",
          round(upper_limit, 2),
          ")")
```

### 3. Find the Critical Value and p-value to interpret the test results.


```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("health_records.csv")

# We want to test whether the average age is 40 years.

# H0 (Null Hypothesis):
# The population mean age is equal to 40 years.

# H1 (Alternative Hypothesis):
# The population mean age is not equal to 40 years.

population_mean = 40


print("HYPOTHESIS TEST")
print("=" * 70)

print("\nH0 (Null Hypothesis):")
print("Mean age = 40 years")

print("\nH1 (Alternative Hypothesis):")
print("Mean age != 40 years")

# SELECT AGE DATA
# Remove missing age values
age_data = df["age"].dropna()

# Sample size
n = len(age_data)

# Sample mean
sample_mean = np.mean(age_data)

# Sample standard deviation
sample_std = np.std(age_data, ddof=1)

# Degrees of freedom
degrees_of_freedom = n - 1

# SET SIGNIFICANCE LEVEL
# Significance level = 5%
alpha = 0.05

# Because this is a two-tailed test,
# we divide alpha by 2.
alpha_two_tail = alpha / 2

# FIND CRITICAL VALUE
# Find the t-critical value
critical_value = stats.t.ppf(
    1 - alpha_two_tail,
    degrees_of_freedom
)

# PERFORM ONE-SAMPLE T-TEST
# Perform one-sample t-test
t_statistic, p_value = stats.ttest_1samp(
    age_data,
    population_mean
)

# DISPLAY RESULTS
print("\n" + "=" * 70)
print("TEST RESULTS")
print("=" * 70)

print("\nSample Size:", n)
print("Sample Mean Age:", round(sample_mean, 2))
print("Sample Standard Deviation:", round(sample_std, 2))
print("Degrees of Freedom:", degrees_of_freedom)

print("\nSignificance Level (Alpha):", alpha)

print("Critical t-value: ±", round(critical_value, 3))

print("Calculated t-statistic:", round(t_statistic, 3))

print("p-value:", round(p_value, 4))


# ============================================================
# STEP 8: INTERPRET THE RESULT
# ============================================================

print("\n" + "=" * 70)
print("INTERPRETATION")
print("=" * 70)

# Method 1: Compare p-value with alpha
if p_value < alpha:
    print("\nDecision: Reject H0")
    print("Because p-value is less than 0.05.")

    print("\nConclusion:")
    print("There is statistically significant evidence that")
    print("the average age is different from 40 years.")

else:
    print("\nDecision: Fail to Reject H0")
    print("Because p-value is greater than or equal to 0.05.")

    print("\nConclusion:")
    print("There is not enough statistical evidence to conclude")
    print("that the average age is different from 40 years.")

# CRITICAL VALUE METHOD
if abs(t_statistic) > critical_value:

    print("\nSince |t-statistic| > Critical Value:")
    print("Reject H0.")

else:

    print("\nSince |t-statistic| <= Critical Value:")
    print("Fail to Reject H0.")
```

### 4. Perform z-test or t-test based on sample size (mean comparison across groups).

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("health_records.csv")

# SELECT TWO GROUPS
# We will compare BMI between:
# Group 1 = Smokers
# Group 2 = Non-Smokers
smokers = df[df["smoking_status"] == "Smoker"]["bmi"].dropna()

non_smokers = df[df["smoking_status"] == "Non-Smoker"]["bmi"].dropna()

# SET HYPOTHESIS
# H0 (Null Hypothesis):
# There is no significant difference between
# the mean BMI of smokers and non-smokers.

# H1 (Alternative Hypothesis):
# There is a significant difference between
# the mean BMI of smokers and non-smokers.

print("\n" + "=" * 70)
print("HYPOTHESIS TEST")
print("=" * 70)

print("\nH0 (Null Hypothesis):")
print("There is no significant difference between the mean")
print("BMI of smokers and non-smokers.")

print("\nH1 (Alternative Hypothesis):")
print("There is a significant difference between the mean")
print("BMI of smokers and non-smokers.")

# CALCULATE GROUP INFORMATION
# Sample size of smokers
n_smokers = len(smokers)

# Sample size of non-smokers
n_non_smokers = len(non_smokers)

# Mean BMI of smokers
mean_smokers = np.mean(smokers)

# Mean BMI of non-smokers
mean_non_smokers = np.mean(non_smokers)

# Standard deviation of smokers
std_smokers = np.std(smokers, ddof=1)

# Standard deviation of non-smokers
std_non_smokers = np.std(non_smokers, ddof=1)

#SET SIGNIFICANCE LEVEL
alpha = 0.05

#PERFORM TWO-SAMPLE T-TEST

# Welch's independent two-sample t-test
# equal_var=False means we do not assume
# that both groups have equal variance.

t_statistic, p_value = stats.ttest_ind(
    smokers,
    non_smokers,
    equal_var=False
)

#FIND DEGREES OF FREEDOM
# Calculate degrees of freedom for Welch's t-test

variance_smokers = std_smokers ** 2
variance_non_smokers = std_non_smokers ** 2

degrees_of_freedom = (
    (variance_smokers / n_smokers +
     variance_non_smokers / n_non_smokers) ** 2
) / (
    ((variance_smokers / n_smokers) ** 2 / (n_smokers - 1))
    +
    ((variance_non_smokers / n_non_smokers) ** 2 / (n_non_smokers - 1))
)

#FIND CRITICAL VALUE
# This is a two-tailed test.
# Therefore, alpha is divided by 2.

critical_value = stats.t.ppf(
    1 - alpha / 2,
    degrees_of_freedom
)

#DISPLAY RESULTS
print("\n" + "=" * 70)
print("TEST RESULTS")
print("=" * 70)

print("\nSMOKERS")
print("Sample Size:", n_smokers)
print("Mean BMI:", round(mean_smokers, 2))
print("Standard Deviation:", round(std_smokers, 2))

print("\nNON-SMOKERS")
print("Sample Size:", n_non_smokers)
print("Mean BMI:", round(mean_non_smokers, 2))
print("Standard Deviation:", round(std_non_smokers, 2))

print("\nDegrees of Freedom:", round(degrees_of_freedom, 2))

print("Significance Level (Alpha):", alpha)

print("Critical t-value: ±", round(critical_value, 3))

print("Calculated t-statistic:", round(t_statistic, 3))

print("p-value:", round(p_value, 4))

# INTERPRET THE RESULT USING p-VALUE
if p_value < alpha:

    print("\nDecision: Reject H0")
    print("Because p-value is less than 0.05.")

    print("\nConclusion:")
    print("There is statistically significant evidence that")
    print("the mean BMI of smokers and non-smokers is different.")

else:

    print("\nDecision: Fail to Reject H0")
    print("Because p-value is greater than or equal to 0.05.")

    print("\nConclusion:")
    print("There is not enough statistical evidence to conclude")
    print("that the mean BMI of smokers and non-smokers is different.")

#INTERPRET THE RESULT USING CRITICAL VALUE
if abs(t_statistic) > critical_value:

    print("\nSince |t-statistic| > Critical Value:")
    print("Reject H0.")

else:

    print("\nSince |t-statistic| <= Critical Value:")
    print("Fail to Reject H0.")
```

### 5. Conduct a chi-square test on categorical data (e.g., Smoking habit vs Disease).


```python
import pandas as pd
from scipy import stats

df = pd.read_csv("health_records.csv")

# CREATE CONTINGENCY TABLE
# Create a table of Smoking Status and Diabetes
table = pd.crosstab(
    df["smoking_status"],
    df["diabetes"]
)

print("CONTINGENCY TABLE")
print("=" * 50)
print(table)

# PERFORM CHI-SQUARE TEST
chi2, p_value, degrees_of_freedom, expected = stats.chi2_contingency(table)

#DISPLAY RESULTS
print("\nCHI-SQUARE TEST RESULTS")
print("=" * 50)

print("Chi-square value:", round(chi2, 3))
print("Degrees of freedom:", degrees_of_freedom)
print("p-value:", round(p_value, 4))

# INTERPRETATION
alpha = 0.05

if p_value < alpha:
    print("\nDecision: Reject H0")
    print("There is a significant relationship between")
    print("smoking status and diabetes.")

else:
    print("\nDecision: Fail to Reject H0")
    print("There is no significant relationship between")
    print("smoking status and diabetes.")

# HYPOTHESES
print("\nH0: Smoking status and diabetes are independent.")
print("H1: Smoking status and diabetes are not independent.")
```

### 6. Perform an ANOVA test to check if age groups significantly differ in disease rate.


```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# 1. Read the CSV file
df = pd.read_csv('health_records.csv')

# 2. Convert diabetes to 0 and 1
df['diabetes_rate'] = df['diabetes'].astype(int)

# 3. Fit the Ordinary Least Squares (OLS) model for One-Way ANOVA
# Aama aapde 'diabetes_rate' ne 'age_group' na adhare compare karie chie
model = ols('diabetes_rate ~ C(age_group)', data=df).fit()

# 4. Perform ANOVA table calculation
anova_table = sm.stats.anova_lm(model, typ=2)

# 5. Display Results
print('=' * 50)
print('ANOVA TEST RESULTS (Using Statsmodels)')
print('=' * 50)
print(anova_table)

# F-statistic ane p-value extract karva
f_statistic = anova_table['F']['C(age_group)']
p_value = anova_table['PR(>F)']['C(age_group)']

print('\nF-statistic:', round(f_statistic, 3))
print('p-value:', round(p_value, 4))

# 6. Interpretation
alpha = 0.05
if p_value < alpha:
  print('\nDecision: Reject H0')
  print('Age groups have significantly different disease rates.')
else:
  print('\nDecision: Fail to Reject H0')
  print('There is no significant difference in disease rates among age groups.')

# 7. Hypotheses
print('\nH0: All age groups have the same diabetes rate.')
print('H1: At least one age group has a different diabetes rate.')
```

### 8. Clearly state the result and interpretation (Accept/Reject H₀) of each test performed.


```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("health_records.csv")

alpha = 0.05

# ONE-SAMPLE T-TEST: AGE

# H0: Mean age = 40
# H1: Mean age != 40

t_age, p_age = stats.ttest_1samp(
    df["age"].dropna(),
    40
)

print("=" * 70)
print("1. ONE-SAMPLE T-TEST: AGE")
print("=" * 70)

print("t-statistic:", round(t_age, 3))
print("p-value:", round(p_age, 4))

if p_age < alpha:
    print("Decision: Reject H0")
    print("Result: Mean age is significantly different from 40.")
else:
    print("Decision: Fail to Reject H0")
    print("Result: Mean age is not significantly different from 40.")

# TWO-SAMPLE T-TEST: SMOKERS vs NON-SMOKERS

# H0: Mean BMI is the same for both groups
# H1: Mean BMI is different for both groups

smokers = df[df["smoking_status"] == "Smoker"]["bmi"].dropna()
non_smokers = df[df["smoking_status"] == "Non-Smoker"]["bmi"].dropna()

t_bmi, p_bmi = stats.ttest_ind(
    smokers,
    non_smokers,
    equal_var=False
)

print("\n" + "=" * 70)
print("2. TWO-SAMPLE T-TEST: BMI")
print("=" * 70)

print("t-statistic:", round(t_bmi, 3))
print("p-value:", round(p_bmi, 4))

if p_bmi < alpha:
    print("Decision: Reject H0")
    print("Result: Mean BMI is significantly different between")
    print("smokers and non-smokers.")
else:
    print("Decision: Fail to Reject H0")
    print("Result: No significant difference in mean BMI between")
    print("smokers and non-smokers.")

# CHI-SQUARE TEST: SMOKING vs DIABETES

# H0: Smoking and diabetes are independent
# H1: Smoking and diabetes are related

table = pd.crosstab(
    df["smoking_status"],
    df["diabetes"]
)

chi2, p_chi, dof, expected = stats.chi2_contingency(table)

print("\n" + "=" * 70)
print("3. CHI-SQUARE TEST: SMOKING vs DIABETES")
print("=" * 70)

print("Chi-square value:", round(chi2, 3))
print("p-value:", round(p_chi, 4))

if p_chi < alpha:
    print("Decision: Reject H0")
    print("Result: Smoking status and diabetes are significantly related.")
else:
    print("Decision: Fail to Reject H0")
    print("Result: No significant relationship between smoking status")
    print("and diabetes.")

# ANOVA: AGE GROUP vs DIABETES RATE

# H0: All age groups have the same diabetes rate
# H1: At least one age group has a different diabetes rate

df["diabetes_rate"] = df["diabetes"].astype(int)

groups = []

for age_group in df["age_group"].unique():
    data = df[df["age_group"] == age_group]["diabetes_rate"]
    groups.append(data)

f_value, p_anova = stats.f_oneway(*groups)

print("\n" + "=" * 70)
print("4. ANOVA: AGE GROUP vs DIABETES RATE")
print("=" * 70)

print("F-statistic:", round(f_value, 3))
print("p-value:", round(p_anova, 4))

if p_anova < alpha:
    print("Decision: Reject H0")
    print("Result: Age groups have significantly different diabetes rates.")
else:
    print("Decision: Fail to Reject H0")
    print("Result: No significant difference in diabetes rates")
    print("among age groups.")

# CORRELATION: AGE vs BMI

# H0: There is no significant correlation between Age and BMI
# H1: There is a significant correlation between Age and BMI

correlation, p_corr = stats.pearsonr(
    df["age"],
    df["bmi"]
)

print("\n" + "=" * 70)
print("5. CORRELATION: AGE vs BMI")
print("=" * 70)

print("Correlation:", round(correlation, 3))
print("p-value:", round(p_corr, 4))

if p_corr < alpha:
    print("Decision: Reject H0")
    print("Result: There is a significant relationship between Age and BMI.")
else:
    print("Decision: Fail to Reject H0")
    print("Result: There is no significant linear relationship")
    print("between Age and BMI.")
```
